```python
import pandas as pd


pmc_df = pd.read_csv('../../assets/pmc_data.csv')
detail_df = pd.read_csv('../../assets/pmc_details.csv')

merged_df = pd.merge(pmc_df, detail_df, on='id', how='left')

cols_to_drop = ['date_published', 'date_updated', 'year_filter', 'category_y', 'scraped_at', 'creator_id']
merged_df.drop(columns=cols_to_drop, inplace=True, errors='ignore')

merged_df.rename(columns={'category_x': 'category', 'creator': 'author'}, inplace=True)
merged_df['category'] = merged_df['category'].str.replace('/', '|')

merged_df['updated_at'] = merged_df['post_date']
merged_df['updated_at'] = merged_df['updated_at'].map(pd.to_datetime)
merged_df.drop(columns=['post_date'], inplace=True)

# merged_df['download_mirrors'] = merged_df['download_mirrors'].map(lambda x: "|".join([val for val in x.split('|') if 'schema' not in val.lower()]))

def parse_k_m(val):
    if not isinstance(val, str):
        return val
    val = val.lower().replace(',', '')
    if 'k' in val:
        return int(float(val.replace('k', '')) * 1_000)
    if 'm' in val:
        return int(float(val.replace('m', '')) * 1_000_000)
    return int(float(val))

metric_cols = ['views', 'downloads', 'diamonds', 'favorites', 'author_subs']
for col in metric_cols:
    merged_df[col] = merged_df[col].apply(parse_k_m)

if 'author_level' in merged_df.columns:
    merged_df['author_level'] = pd.to_numeric(merged_df['author_level'], errors='coerce').fillna(0).astype(int)

ordered_columns = [
    # Identifiers & Core Info
    'id', 'platform', 'title', 'category', 'url',
    
    # Author Stats
    'author_id', 'author', 'author_rank', 'author_level', 'author_subs',
    
    # Popularity Metrics
    'views', 'downloads', 'favorites', 'diamonds', 'comments',
    
    # Content & Links
    'tags', 'download_mirrors', 'description', 'gallery_urls',
    
    # Timeline
    'updated_at'
]

merged_df = merged_df[ordered_columns]

to_int = ['author_id', 'author_subs', 'comments', 'views', 'downloads', 'favorites', 'diamonds']
for col in to_int:
    merged_df[col] = pd.to_numeric(merged_df[col], errors='coerce').fillna(0).astype(int)

to_str = ['title', 'category', 'url', 'author_id', 'author', 'tags', 'description', 'download_mirrors', 'gallery_urls', 'author_rank']
for col in to_str:
    merged_df[col] = merged_df[col].astype(pd.StringDtype())

print(merged_df.dtypes)
merged_df.head(100)
```

```python
merged_df["download_mirrors"] = merged_df["download_mirrors"].fillna("").apply(lambda x: "|".join([val for val in x.split('|') if 'schem' not in val.lower()]))
```
